```python
import numpy as np
import pandas as pd
```

```python
#create a dtat frame by the name movies
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
#merging both data frames
movies = movies.merge(credits,on='title')
```

```python
movies.head(1)
```

```python
#genres
#id
#keywords
#title
#overview
#cast 
#crew
# removing only these coloum from the data
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.info()
```

```python
#the dataframe on which we will work
movies.head()
```

```python
# misssing data
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
#dublicate data
movies.duplicated().sum()
```

```python
#iloc is used to select rows and columns from a DataFrame using their index positions
movies.iloc[0].genres
```

```python
#create a helper function(A helper function is a small function that helps another function perform its task)
import ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L


```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
def convert3(obj):
    L = []
    counter =0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter+=1
        else:
            break
    return L
```

```python
movies['cast']=movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
import ast

def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L

 
```

```python
movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
#(lambda x: x.split()) is a lambda function in Python. It is a short way to write a function in one line.
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
#removing all the spaces between words
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","")for i in x])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","")for i in x])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","")for i in x])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","")for i in x])
```

```python
movies.head()
```

```python
#concatination all the things into tags
movies['tags']=movies['overview'] + movies['keywords'] + movies['crew'] + movies['cast'] + movies['genres']
```

```python
movies.head()
```

```python
new_df=movies[['movie_id','title','tags']]
```

```python
new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df.head()
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
       y.append( ps.stem(i))

    return" ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

```python
new_df['tags'][0]
```

```python
#convert into lower case
new_df['tags'] = new_df['tags'].apply(
    lambda x: [word.lower() for word in x] if isinstance(x, list) else x.lower()
)
```

```python
new_df.head()
```

```python
new_df['tags'] = new_df['tags'].apply(
    lambda x: ' '.join(map(str, x)).lower()
    if isinstance(x, list) else str(x).lower()
)
```

```python
#text vectorization
from sklearn.feature_extraction.text import CountVectorizer
cv =CountVectorizer(max_features = 5000,stop_words = 'english')
```

```python
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title']==movie].index[0]
    distances = similarity[movie_index]
    movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(i)
    return
```
